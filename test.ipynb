```python
import cv2
import time
import numpy as np


 
rtsp_url1 = "rtsp://admin:Doer2022%24%23@202.125.77.226:554/Streaming/Channels/101"
rtsp_url2 = "rtsp://admin:Doer2022%24%23@202.125.77.226:554/Streaming/Channels/501"

cap1 = cv2.VideoCapture(rtsp_url1, cv2.CAP_FFMPEG)
cap2 = cv2.VideoCapture(rtsp_url2, cv2.CAP_FFMPEG)

# Reduce buffer (important for latency)
cap1.set(cv2.CAP_PROP_BUFFERSIZE, 1)
cap2.set(cv2.CAP_PROP_BUFFERSIZE, 1)

prev_time = time.time()

while True:
    ret1, frame1 = cap1.read()
    ret2, frame2 = cap2.read()

    if not ret1 or not ret2:
        print("Failed to grab frame")
        break

    # Resize to same size (important for grid)
    frame1 = cv2.resize(frame1, (640, 360))
    frame2 = cv2.resize(frame2, (640, 360))

    # Combine side by side
    grid = np.hstack((frame1, frame2))

    # FPS
    curr_time = time.time()
    fps = 1 / (curr_time - prev_time)
    prev_time = curr_time

    cv2.putText(grid, f"FPS: {fps:.2f}", (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    cv2.imshow("RTSP Grid View", grid)

    if cv2.waitKey(1) == 27:
        break

cap1.release()
cap2.release()
cv2.destroyAllWindows()
```
